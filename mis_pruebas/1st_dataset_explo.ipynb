## Meses en que aparece cada cliente

```python
cliente_mes = df.groupby("numero_de_cliente")["foto_mes"].apply(list)
cliente_mes.reset_index()
```

```python
for i,row in cliente_mes.reset_index().iterrows():
    # busquemos cosas raras
    # print(type(row['foto_mes']))
    # row['foto_mes'].sort()
    # print(row['foto_mes'])
    for i in range(len(row['foto_mes'])-1):
        cliente = row['numero_de_cliente']
        meses = row['foto_mes']
        # desordenados
        if row['foto_mes'][i]>row['foto_mes'][i+1]:
            print(f'MESES DESORDENADOS, cliente: {cliente}, meses: {meses}')
        # espacios en blanco
        if (row['foto_mes'][i+1]-row['foto_mes'][i])>1:
                    print(f'GAPS EN MESES, cliente: {cliente}, meses: {meses}')
        # meses duplicados
        if len(meses) != len(set(meses)):
              print(f'MESES DUPLICADOS, cliente: {cliente}, meses: {meses}')
```

```python
def ternaria(row, log=False):
    if log :print('-----------------------')
    # Numero de cliente y meses en q esta presente
    cliente = row['numero_de_cliente']
    mes_registro = row['foto_mes']
    # cliente_mes= cliente_mes.reset_index()
    meses_presente= cliente_mes.loc[cliente_mes['numero_de_cliente']==cliente]['foto_mes'].values[0]
    if log :print('mes registro', mes_registro)
    if log :print('cliente', cliente)
    if log :print('meses presente' ,meses_presente)
    if log :print(type(meses_presente))
    # casos limite
    # 1 -> estamos parados en 202108 -> no tenemos info para adelante
    if mes_registro == 202108:
        if log :print('caso1')
        return np.nan
    # 2 -> estamos parados en 202107 -> si 202108 no existe = baja+1 else  'EOS'
    if mes_registro == 202107:
        if log : print('caso2')
        if log : print(202108 in meses_presente)
        return 'baja+1' if 202108 not in meses_presente else np.nan
    # 3 -> caso general 
    if log :print('caso general')
    if mes_registro+1 in meses_presente and mes_registro+2 in meses_presente:
        if log : print('continua')
        return 'continua'
    if mes_registro +1 in meses_presente:
        if log : print('baja+2')
        return 'baja+2'
    if mes_registro+1 not in meses_presente:
        if log : print('baja+1')
        return 'baja+1'
```

```python
df['ternaria'] = df.apply(ternaria,axis=1)
```

```python
df.ternaria.value_counts(dropna=False)
```
